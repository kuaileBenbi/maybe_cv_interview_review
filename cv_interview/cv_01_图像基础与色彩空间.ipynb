```python
# ====== 参考答案 ======

def rgb_to_hsv_manual(rgb_img):
    """手动实现 RGB -> HSV 转换"""
    # 归一化到 [0, 1]
    img_float = rgb_img.astype(np.float64) / 255.0
    R, G, B = img_float[:,:,0], img_float[:,:,1], img_float[:,:,2]
    
    # 计算 V, min_val
    V = np.maximum(np.maximum(R, G), B)
    min_val = np.minimum(np.minimum(R, G), B)
    diff = V - min_val
    
    # 计算 S
    S = np.where(V > 0, diff / V, 0.0)
    
    # 计算 H
    H = np.zeros_like(V)
    
    # 当 diff == 0 时, H = 0 (灰色)
    mask = diff > 0
    
    # V == R
    m = mask & (V == R)
    H[m] = 60.0 * (((G[m] - B[m]) / diff[m]) % 6)
    
    # V == G
    m = mask & (V == G)
    H[m] = 60.0 * (((B[m] - R[m]) / diff[m]) + 2)
    
    # V == B
    m = mask & (V == B)
    H[m] = 60.0 * (((R[m] - G[m]) / diff[m]) + 4)
    
    # 缩放到 OpenCV 范围
    H_out = (H / 2.0).astype(np.uint8)        # 0-180
    S_out = (S * 255).astype(np.uint8)          # 0-255
    V_out = (V * 255).astype(np.uint8)          # 0-255
    
    hsv = np.stack([H_out, S_out, V_out], axis=-1)
    return hsv

# 验证
test = np.array([[[255, 0, 0],
                   [0, 255, 0],
                   [0, 0, 255],
                   [255, 255, 0],
                   [0, 255, 255]]], dtype=np.uint8)

hsv_manual = rgb_to_hsv_manual(test)
hsv_cv2 = cv2.cvtColor(test, cv2.COLOR_RGB2HSV)
print('手动转换结果:\n', hsv_manual)
print('OpenCV结果:\n', hsv_cv2)
print('差异:', np.abs(hsv_manual.astype(int) - hsv_cv2.astype(int)).max(), '(应接近0)')
```

```python
# ====== 参考答案 ======

def histogram_equalization(gray_img):
    """手动实现直方图均衡化"""
    # 1. 计算直方图
    hist = np.zeros(256, dtype=np.int64)
    for val in gray_img.flatten():
        hist[val] += 1
    
    # 更高效的写法: hist = np.bincount(gray_img.flatten(), minlength=256)
    
    # 2. 计算累积分布函数 (CDF)
    cdf = hist.cumsum()
    
    # 3. 归一化 CDF 到 [0, 255]
    # 排除 cdf=0 的情况
    cdf_min = cdf[cdf > 0].min()
    total = gray_img.size
    cdf_normalized = ((cdf - cdf_min) / (total - cdf_min) * 255).astype(np.uint8)
    
    # 4. 用 CDF 作为查找表映射
    equalized = cdf_normalized[gray_img]
    
    return equalized

# 测试
np.random.seed(42)
dark_img = np.random.randint(0, 100, (200, 300), dtype=np.uint8)

result_manual = histogram_equalization(dark_img)
result_cv2 = cv2.equalizeHist(dark_img)

print(f'原图均值: {dark_img.mean():.1f}')
print(f'手动均衡化均值: {result_manual.mean():.1f}')
print(f'OpenCV均衡化均值: {result_cv2.mean():.1f}')

# 可视化对比
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].imshow(dark_img, cmap='gray')
axes[0, 0].set_title('原图')
axes[0, 1].imshow(result_manual, cmap='gray')
axes[0, 1].set_title('手动均衡化')
axes[1, 0].hist(dark_img.flatten(), bins=256, range=(0, 256))
axes[1, 0].set_title('原图直方图')
axes[1, 1].hist(result_manual.flatten(), bins=256, range=(0, 256))
axes[1, 1].set_title('均衡化后直方图')
plt.tight_layout()
plt.show()
```

```python
# ====== 参考答案 ======

def rotate_image(img, angle):
    """图像绕中心旋转, 不裁剪, 白色填充"""
    h, w = img.shape[:2]
    cx, cy = w / 2, h / 2
    
    # 计算旋转矩阵
    M = cv2.getRotationMatrix2D((cx, cy), angle, 1.0)
    
    # 计算新图像尺寸 (确保不裁剪)
    cos_val = np.abs(M[0, 0])
    sin_val = np.abs(M[0, 1])
    new_w = int(h * sin_val + w * cos_val)
    new_h = int(h * cos_val + w * sin_val)
    
    # 调整旋转中心
    M[0, 2] += (new_w - w) / 2
    M[1, 2] += (new_h - h) / 2
    
    rotated = cv2.warpAffine(img, M, (new_w, new_h), borderValue=(255, 255, 255))
    return rotated

# 测试
test_img = np.zeros((200, 300, 3), dtype=np.uint8)
cv2.rectangle(test_img, (100, 50), (200, 150), (0, 255, 0), 2)
cv2.putText(test_img, 'OpenCV', (100, 110), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for i, angle in enumerate([0, 30, 45, 90]):
    rotated = rotate_image(test_img, angle)
    axes[i].imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
    axes[i].set_title(f'旋转 {angle}度')
    axes[i].axis('off')
plt.tight_layout()
plt.show()
```
