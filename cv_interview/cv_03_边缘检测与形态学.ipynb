```python
# ====== 参考答案 ======

def remove_noise_morphology(binary_img):
    """用形态学操作去除二值图像噪点"""
    # 使用开运算去除白色小噪点 (先腐蚀后膨胀)
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
    opened = cv2.morphologyEx(binary_img, cv2.MORPH_OPEN, kernel)
    
    # 使用闭运算填充黑色小孔洞 (先膨胀后腐蚀)
    closed = cv2.morphologyEx(opened, cv2.MORPH_CLOSE, kernel)
    
    return closed

# 生成测试图像
np.random.seed(42)
clean = np.zeros((200, 300), dtype=np.uint8)
cv2.rectangle(clean, (50, 50), (150, 150), 255, -1)
cv2.circle(clean, (230, 100), 40, 255, -1)

noise = clean.copy()
noise_mask = np.random.random(noise.shape) < 0.05
noise[noise_mask] = 255 - noise[noise_mask]

result = remove_noise_morphology(noise)

# 可视化
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(clean, cmap='gray'); axes[0].set_title('原始图像')
axes[1].imshow(noise, cmap='gray'); axes[1].set_title(f'加噪图像 (差异: {np.sum(noise != clean)} 像素)')
axes[2].imshow(result, cmap='gray'); axes[2].set_title(f'去噪后 (残余: {np.sum(result != clean)} 像素)')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
# ====== 参考答案 ======

def canny_manual(gray_img, low_threshold, high_threshold):
    """手动实现 Canny 边缘检测"""
    
    # Step 1: 高斯平滑
    blurred = cv2.GaussianBlur(gray_img, (5, 5), 1.4)
    
    # Step 2: 计算梯度 (Sobel)
    grad_x = cv2.Sobel(blurred, cv2.CV_64F, 1, 0, ksize=3)
    grad_y = cv2.Sobel(blurred, cv2.CV_64F, 0, 1, ksize=3)
    
    # 梯度幅值和方向
    magnitude = np.sqrt(grad_x**2 + grad_y**2)
    direction = np.arctan2(grad_y, grad_x) * 180 / np.pi  # 转为角度
    direction[direction < 0] += 180  # 映射到 [0, 180)
    
    h, w = gray_img.shape
    
    # Step 3: 非极大值抑制 (NMS)
    nms = np.zeros_like(magnitude)
    for i in range(1, h-1):
        for j in range(1, w-1):
            angle = direction[i, j]
            
            # 根据梯度方向选取两个相邻像素
            if (0 <= angle < 22.5) or (157.5 <= angle <= 180):
                # 水平方向: 左右
                n1, n2 = magnitude[i, j-1], magnitude[i, j+1]
            elif 22.5 <= angle < 67.5:
                # 对角线: 左上-右下
                n1, n2 = magnitude[i-1, j-1], magnitude[i+1, j+1]
            elif 67.5 <= angle < 112.5:
                # 垂直方向: 上下
                n1, n2 = magnitude[i-1, j], magnitude[i+1, j]
            else:  # 112.5 <= angle < 157.5
                # 对角线: 右上-左下
                n1, n2 = magnitude[i-1, j+1], magnitude[i+1, j-1]
            
            # 仅保留局部最大值
            if magnitude[i, j] >= n1 and magnitude[i, j] >= n2:
                nms[i, j] = magnitude[i, j]
    
    # Step 4: 双阈值检测与边缘连接
    strong = 255
    weak = 75
    
    result = np.zeros_like(nms, dtype=np.uint8)
    result[nms >= high_threshold] = strong
    result[(nms >= low_threshold) & (nms < high_threshold)] = weak
    
    # 边缘连接: 弱边缘若与强边缘相邻则保留
    for i in range(1, h-1):
        for j in range(1, w-1):
            if result[i, j] == weak:
                # 检查8邻域是否有强边缘
                if np.any(result[i-1:i+2, j-1:j+2] == strong):
                    result[i, j] = strong
                else:
                    result[i, j] = 0
    
    return result

# 测试并对比
test_img = np.zeros((100, 100), dtype=np.uint8)
test_img[20:80, 20:80] = 200
test_img[40:60, 40:60] = 100

edges_manual = canny_manual(test_img, 50, 150)
edges_cv2 = cv2.Canny(test_img, 50, 150)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(test_img, cmap='gray'); axes[0].set_title('原图')
axes[1].imshow(edges_manual, cmap='gray'); axes[1].set_title(f'手动 Canny (边缘: {np.sum(edges_manual>0)})')
axes[2].imshow(edges_cv2, cmap='gray'); axes[2].set_title(f'OpenCV Canny (边缘: {np.sum(edges_cv2>0)})')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
# ====== 参考答案 ======

def detect_and_filter_contours(binary_img, min_area=500):
    """检测轮廓并筛选"""
    # 查找轮廓
    contours, hierarchy = cv2.findContours(
        binary_img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
    )
    
    filtered_contours = []
    info = []
    
    for cnt in contours:
        area = cv2.contourArea(cnt)
        if area >= min_area:
            perimeter = cv2.arcLength(cnt, True)
            x, y, w, h = cv2.boundingRect(cnt)
            
            filtered_contours.append(cnt)
            info.append({
                'area': area,
                'perimeter': perimeter,
                'bbox': (x, y, w, h),
                'aspect_ratio': float(w) / h if h > 0 else 0,
                'circularity': 4 * np.pi * area / (perimeter**2) if perimeter > 0 else 0
            })
    
    return filtered_contours, info

# 生成测试图像
test = np.zeros((400, 600), dtype=np.uint8)
cv2.rectangle(test, (30, 30), (130, 130), 255, -1)
cv2.circle(test, (300, 80), 50, 255, -1)
cv2.rectangle(test, (450, 30), (470, 50), 255, -1)
cv2.circle(test, (100, 300), 20, 255, -1)
cv2.rectangle(test, (250, 250), (260, 260), 255, -1)
cv2.ellipse(test, (450, 300), (60, 30), 0, 0, 360, 255, -1)

contours, info = detect_and_filter_contours(test, min_area=500)

# 可视化
result = cv2.cvtColor(test, cv2.COLOR_GRAY2BGR)
colors = [(0, 255, 0), (0, 0, 255), (255, 0, 0), (255, 255, 0), (0, 255, 255)]

print(f'检测到 {len(contours)} 个面积>=500 的轮廓:\n')
for i, (cnt, inf) in enumerate(zip(contours, info)):
    color = colors[i % len(colors)]
    cv2.drawContours(result, [cnt], -1, color, 2)
    x, y, w, h = inf['bbox']
    cv2.rectangle(result, (x, y), (x+w, y+h), color, 1)
    cv2.putText(result, f'#{i} A={inf["area"]:.0f}', (x, y-5),
                cv2.FONT_HERSHEY_SIMPLEX, 0.4, color, 1)
    print(f'  轮廓#{i}: 面积={inf["area"]:.0f}, 周长={inf["perimeter"]:.0f}, '
          f'外接矩形={inf["bbox"]}, 宽高比={inf["aspect_ratio"]:.2f}, 圆度={inf["circularity"]:.3f}')

plt.figure(figsize=(12, 6))
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title('轮廓检测结果 (绿色框=外接矩形)')
plt.axis('off')
plt.show()
```

```python
# ====== 参考答案 ======

def locate_plate_regions(img):
    """使用形态学操作定位车牌候选区域"""
    # 1. 灰度化
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    # 2. Sobel 边缘检测 (提取垂直边缘, 车牌字符有丰富的垂直边缘)
    sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sobel_x = np.abs(sobel_x).astype(np.uint8)
    
    # 3. 二值化
    _, binary = cv2.threshold(sobel_x, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    
    # 4. 闭运算: 先膨胀连接相邻字符, 再腐蚀恢复大小
    # 使用宽而矮的矩形核, 匹配车牌形状
    kernel_close = cv2.getStructuringElement(cv2.MORPH_RECT, (17, 5))
    closed = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel_close)
    
    # 5. 开运算: 去除小噪点
    kernel_open = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
    opened = cv2.morphologyEx(closed, cv2.MORPH_OPEN, kernel_open)
    
    # 6. 查找轮廓
    contours, _ = cv2.findContours(opened, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    # 7. 筛选: 面积 + 宽高比 (车牌宽高比通常在 2~5 之间)
    candidates = []
    result_img = img.copy()
    
    for cnt in contours:
        area = cv2.contourArea(cnt)
        x, y, w, h = cv2.boundingRect(cnt)
        aspect_ratio = float(w) / h if h > 0 else 0
        
        # 筛选条件
        if area > 1000 and 2.0 <= aspect_ratio <= 6.0 and w > 60 and h > 20:
            candidates.append((x, y, w, h))
            cv2.rectangle(result_img, (x, y), (x+w, y+h), (0, 255, 0), 2)
            cv2.putText(result_img, f'AR={aspect_ratio:.1f}', (x, y-5),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)
    
    return candidates, result_img

# 生成模拟车牌图像
img = np.full((300, 500, 3), 120, dtype=np.uint8)
np.random.seed(42)
for _ in range(20):
    x, y = np.random.randint(0, 500), np.random.randint(0, 300)
    cv2.circle(img, (x, y), np.random.randint(5, 20),
               (np.random.randint(80, 160),)*3, -1)

plate_x, plate_y = 150, 120
cv2.rectangle(img, (plate_x, plate_y), (plate_x+200, plate_y+60), (255, 255, 255), -1)
cv2.rectangle(img, (plate_x, plate_y), (plate_x+200, plate_y+60), (0, 0, 200), 2)
for i, ch in enumerate('A B 1 2 3'.split()):
    cv2.putText(img, ch, (plate_x+15+i*35, plate_y+42),
                cv2.FONT_HERSHEY_SIMPLEX, 1.0, (0, 0, 0), 2)

candidates, result = locate_plate_regions(img)

print(f'找到 {len(candidates)} 个候选区域')
for i, (x, y, w, h) in enumerate(candidates):
    print(f'  候选{i}: ({x}, {y}, {w}, {h}), 宽高比={w/h:.2f}')

# 可视化处理流程
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
sobel_x = np.abs(cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)).astype(np.uint8)
_, binary = cv2.threshold(sobel_x, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes[0,0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); axes[0,0].set_title('1. 原图')
axes[0,1].imshow(sobel_x, cmap='gray'); axes[0,1].set_title('2. Sobel X 梯度')
axes[0,2].imshow(binary, cmap='gray'); axes[0,2].set_title('3. 二值化')

kernel_close = cv2.getStructuringElement(cv2.MORPH_RECT, (17, 5))
closed = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel_close)
kernel_open = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
opened = cv2.morphologyEx(closed, cv2.MORPH_OPEN, kernel_open)

axes[1,0].imshow(closed, cmap='gray'); axes[1,0].set_title('4. 闭运算 (连接字符)')
axes[1,1].imshow(opened, cmap='gray'); axes[1,1].set_title('5. 开运算 (去噪)')
axes[1,2].imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB)); axes[1,2].set_title('6. 候选区域')

for ax in axes.flat:
    ax.axis('off')
plt.tight_layout()
plt.show()
```

| 知识点 | 要点 |
|--------|------|
| Canny 四步骤 | 高斯 → Sobel → NMS → 双阈值 |
| Canny 双阈值 | 强边缘必留, 弱边缘看连接, 推荐比 2:1~3:1 |
| NMS 作用 | 沿梯度方向比较, 细化边缘为单像素 |

| 腐蚀 | 白区缩小, 去小白点, 取最小值 |
| 开运算 | 先腐蚀后膨胀 → 去小物体 |
| 闭运算 | 先膨胀后腐蚀 → 填充孔洞 |
| 形态学梯度 | 膨胀 - 腐蚀 → 提取边缘 |
| findContours | OpenCV 4.x 返回 (contours, hierarchy) |
| 轮廓特征 | contourArea, arcLength, boundingRect |
| 车牌定位思路 | 边缘 → 闭运算连接 → 开运算去噪 → 轮廓筛选宽高比 |
