```python
# ====== 参考答案 ======
import torch

# 生成数据: y = 3*x + 2 + noise
torch.manual_seed(42)
x = torch.randn(100, 1)
y_true = 3 * x + 2 + 0.5 * torch.randn(100, 1)

# 初始化参数
w = torch.randn(1, 1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

learning_rate = 0.1
losses = []

for epoch in range(100):
    # 前向传播
    y_pred = x @ w + b
    
    # 计算损失 (MSE)
    loss = ((y_pred - y_true) ** 2).mean()
    losses.append(loss.item())
    
    # 反向传播
    loss.backward()
    
    # 更新参数 (梯度下降)
    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad
    
    # 清零梯度
    w.grad.zero_()
    b.grad.zero_()

print(f"学到的参数: w={w.item():.4f}, b={b.item():.4f}")
print(f"真实参数:   w=3.0000, b=2.0000")
```

```python
# ====== 参考答案 ======
import torch
import matplotlib.pyplot as plt
import numpy as np

def analyze_gradients(input_dim=10, hidden_dim=50, num_layers=8, 
                      activation='sigmoid', num_trials=100):
    """
    分析不同激活函数下的梯度分布
    """
    activation_fns = {
        'sigmoid': torch.sigmoid,
        'tanh': torch.tanh,
        'relu': torch.relu,
    }
    act_fn = activation_fns[activation]
    
    grad_norms_all = []
    
    for trial in range(num_trials):
        # 创建多层权重
        weights = []
        for i in range(num_layers):
            fan_in = input_dim if i == 0 else hidden_dim
            fan_out = hidden_dim
            w = torch.randn(fan_in, fan_out, requires_grad=True) * 0.1
            weights.append(w)
        
        # 前向传播
        x = torch.randn(1, input_dim)
        h = x
        for w in weights:
            h = act_fn(h @ w)
        
        # 计算损失并反向传播
        loss = h.sum()
        loss.backward()
        
        # 记录每层梯度范数
        grad_norms = [w.grad.norm().item() for w in weights]
        grad_norms_all.append(grad_norms)
    
    # 平均梯度范数
    avg_norms = np.mean(grad_norms_all, axis=0)
    return avg_norms

# 对比不同激活函数
activations = ['sigmoid', 'tanh', 'relu']
plt.figure(figsize=(10, 6))
for act in activations:
    norms = analyze_gradients(activation=act, num_trials=50)
    plt.semilogy(range(len(norms)), norms, 'o-', label=act, linewidth=2)

plt.xlabel('层编号')
plt.ylabel('梯度范数 (log scale)')
plt.title('不同激活函数的梯度流动')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("""结论:
- sigmoid: 梯度随层数指数衰减 (梯度消失), 浅层几乎收不到梯度
- tanh: 比 sigmoid 好, 但深层仍有梯度消失
- relu: 梯度在正区间恒为1, 缓解了梯度消失问题""")
```

```python
# ====== 参考答案 ======
import torch
import time

def benchmark_matmul(device, sizes=[256, 512, 1024, 2048], num_warmup=10, num_iter=100):
    """
    对比不同尺寸矩阵乘法在 CPU 和 GPU 上的速度
    """
    results = {}
    
    for size in sizes:
        # 创建矩阵
        a = torch.randn(size, size, device=device)
        b = torch.randn(size, size, device=device)
        
        # 预热
        for _ in range(num_warmup):
            _ = a @ b
        
        # 如果是 CUDA, 同步后计时
        if device == 'cuda':
            torch.cuda.synchronize()
        
        start = time.perf_counter()
        for _ in range(num_iter):
            c = a @ b
        
        if device == 'cuda':
            torch.cuda.synchronize()
        
        elapsed = time.perf_counter() - start
        avg_ms = elapsed / num_iter * 1000
        results[size] = avg_ms
        print(f"  [{device.upper()}] 矩阵 {size}x{size}: {avg_ms:.3f} ms/iter")
    
    return results

# CPU 基准测试
print("=== CPU 基准测试 ===")
cpu_results = benchmark_matmul('cpu', sizes=[256, 512, 1024])

# GPU 基准测试 (如果可用)
if torch.cuda.is_available():
    print("\n=== GPU 基准测试 ===")
    gpu_results = benchmark_matmul('cuda', sizes=[256, 512, 1024, 2048, 4096])
    
    print("\n=== 加速比 ===")
    for size in cpu_results:
        if size in gpu_results:
            speedup = cpu_results[size] / gpu_results[size]
            print(f"  矩阵 {size}x{size}: GPU 比 CPU 快 {speedup:.1f}x")
else:
    print("\nCUDA 不可用, 跳过 GPU 测试")
    print("提示: 在有 NVIDIA GPU 的环境中运行可以看到显著加速")
```
