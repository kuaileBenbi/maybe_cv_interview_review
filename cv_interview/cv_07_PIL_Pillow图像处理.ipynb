```python
# ====== 参考答案 ======
from PIL import Image, ImageEnhance, ImageOps
import numpy as np
import random

class ImagePipeline:
    """图片预处理流水线"""
    
    def __init__(self, target_size=224, crop_size=224, 
                 random_flip=True, color_jitter=True):
        self.target_size = target_size
        self.crop_size = crop_size
        self.random_flip = random_flip
        self.color_jitter = color_jitter
    
    def resize(self, img):
        """等比缩放, 使短边等于 target_size"""
        w, h = img.size
        if w < h:
            new_w = self.target_size
            new_h = int(h * self.target_size / w)
        else:
            new_h = self.target_size
            new_w = int(w * self.target_size / h)
        return img.resize((new_w, new_h), Image.LANCZOS)
    
    def center_crop(self, img):
        """中心裁剪到 crop_size x crop_size"""
        w, h = img.size
        left = (w - self.crop_size) // 2
        upper = (h - self.crop_size) // 2
        right = left + self.crop_size
        lower = upper + self.crop_size
        return img.crop((left, upper, right, lower))
    
    def random_horizontal_flip(self, img):
        """以 50% 概率水平翻转"""
        if random.random() > 0.5:
            return img.transpose(Image.FLIP_LEFT_RIGHT)
        return img
    
    def apply_color_jitter(self, img, brightness=0.2, contrast=0.2, saturation=0.2):
        """随机颜色抖动"""
        if brightness > 0:
            factor = 1.0 + random.uniform(-brightness, brightness)
            img = ImageEnhance.Brightness(img).enhance(factor)
        if contrast > 0:
            factor = 1.0 + random.uniform(-contrast, contrast)
            img = ImageEnhance.Contrast(img).enhance(factor)
        if saturation > 0:
            factor = 1.0 + random.uniform(-saturation, saturation)
            img = ImageEnhance.Color(img).enhance(factor)
        return img
    
    def __call__(self, img):
        """执行完整流水线"""
        img = self.resize(img)
        img = self.center_crop(img)
        if self.random_flip:
            img = self.random_horizontal_flip(img)
        if self.color_jitter:
            img = self.apply_color_jitter(img)
        return img

# 测试
test_img = Image.fromarray(np.random.randint(0, 256, (300, 400, 3), dtype=np.uint8))
pipeline = ImagePipeline(target_size=256, crop_size=224)
result = pipeline(test_img)
print(f"原图尺寸: {test_img.size}, 输出尺寸: {result.size}")
```

```python
# ====== 参考答案 ======
from PIL import Image, ImageDraw, ImageFont
import numpy as np

# 默认颜色表
DEFAULT_COLORS = [
    (255, 0, 0), (0, 255, 0), (0, 0, 255),
    (255, 255, 0), (0, 255, 255), (255, 0, 255),
    (128, 0, 0), (0, 128, 0), (0, 0, 128),
]

def draw_detections(image, boxes, labels, scores=None, 
                    class_colors=None, thickness=3, font_size=20):
    """
    在图像上绘制检测框和标签
    """
    img = image.copy()
    draw = ImageDraw.Draw(img)
    
    # 尝试加载字体, 失败则用默认字体
    try:
        font = ImageFont.truetype("arial.ttf", font_size)
    except (IOError, OSError):
        font = ImageFont.load_default()
    
    for i, (box, label) in enumerate(zip(boxes, labels)):
        # 确定颜色
        if class_colors and label in class_colors:
            color = class_colors[label]
        else:
            color = DEFAULT_COLORS[i % len(DEFAULT_COLORS)]
        
        x1, y1, x2, y2 = box
        
        # 绘制边界框
        draw.rectangle([x1, y1, x2, y2], outline=color, width=thickness)
        
        # 构建标签文本
        text = label
        if scores is not None:
            text += f" {scores[i]:.2f}"
        
        # 计算文本背景区域
        text_bbox = draw.textbbox((x1, y1), text, font=font)
        text_w = text_bbox[2] - text_bbox[0]
        text_h = text_bbox[3] - text_bbox[1]
        
        # 绘制文本背景和文字
        draw.rectangle([x1, y1 - text_h - 4, x1 + text_w + 4, y1], 
                       fill=color)
        draw.text((x1 + 2, y1 - text_h - 2), text, fill="white", font=font)
    
    return img

# 测试
test_img = Image.fromarray(np.random.randint(100, 200, (400, 600, 3), dtype=np.uint8))
boxes = [[50, 30, 250, 200], [300, 100, 500, 350]]
labels = ["cat", "dog"]
scores = [0.95, 0.87]
result = draw_detections(test_img, boxes, labels, scores)
print(f"标注图尺寸: {result.size}")
```

```python
# ====== 参考答案 ======
import torch
import numpy as np
from PIL import Image
import random

class RandomErasing:
    """随机擦除 transform (PIL 版本)"""
    def __init__(self, p=0.5, scale=(0.02, 0.33), ratio=(0.3, 3.3)):
        self.p = p
        self.scale = scale
        self.ratio = ratio
    
    def __call__(self, img):
        if random.random() > self.p:
            return img
        
        w, h = img.size
        area = w * h
        
        for _ in range(10):  # 最多尝试10次
            target_area = random.uniform(self.scale[0], self.scale[1]) * area
            aspect_ratio = random.uniform(self.ratio[0], self.ratio[1])
            
            erase_w = int(round((target_area * aspect_ratio) ** 0.5))
            erase_h = int(round((target_area / aspect_ratio) ** 0.5))
            
            if erase_w < w and erase_h < h:
                x1 = random.randint(0, w - erase_w)
                y1 = random.randint(0, h - erase_h)
                
                arr = np.array(img)
                arr[y1:y1+erase_h, x1:x1+erase_w, :] = 0
                return Image.fromarray(arr)
        
        return img

class Cutout:
    """Cutout 数据增强"""
    def __init__(self, n_holes=1, length=16):
        self.n_holes = n_holes
        self.length = length
    
    def __call__(self, img):
        arr = np.array(img)
        h, w = arr.shape[:2]
        
        for _ in range(self.n_holes):
            y = random.randint(0, h - 1)
            x = random.randint(0, w - 1)
            
            y1 = max(0, y - self.length // 2)
            y2 = min(h, y + self.length // 2)
            x1 = max(0, x - self.length // 2)
            x2 = min(w, x + self.length // 2)
            
            arr[y1:y2, x1:x2, :] = 0
        
        return Image.fromarray(arr)

class GaussianNoise:
    """高斯噪声 (tensor 版本, 在 ToTensor 之后使用)"""
    def __init__(self, mean=0.0, std=0.1):
        self.mean = mean
        self.std = std
    
    def __call__(self, tensor):
        noise = torch.randn_like(tensor) * self.std + self.mean
        return torch.clamp(tensor + noise, 0.0, 1.0)

# 测试
test_img = Image.fromarray(np.random.randint(0, 256, (100, 100, 3), dtype=np.uint8))
erasing = RandomErasing(p=1.0)
result = erasing(test_img)
print(f"随机擦除: 原图和结果是否相同: {np.array_equal(np.array(test_img), np.array(result))}")

cutout = Cutout(n_holes=2, length=20)
result2 = cutout(test_img)
print(f"Cutout: 原图和结果是否相同: {np.array_equal(np.array(test_img), np.array(result2))}")

noise = GaussianNoise(std=0.1)
test_tensor = torch.rand(3, 100, 100)
noisy = noise(test_tensor)
print(f"高斯噪声: 原图和结果是否相同: {torch.equal(test_tensor, noisy)}")
print(f"噪声范围: [{noisy.min():.3f}, {noisy.max():.3f}]")
```

```python
# ====== 参考答案 ======
from PIL import Image
import numpy as np

def blend_images(img1, img2, alpha=0.5):
    """两张等大图片的线性混合"""
    assert img1.size == img2.size, "两张图片尺寸必须相同"
    assert img1.mode == img2.mode, "两张图片模式必须相同"
    return Image.blend(img1, img2, alpha)

def paste_with_mask(background, foreground, position, mask=None):
    """将前景图粘贴到背景图的指定位置"""
    result = background.copy()
    if mask is not None:
        result.paste(foreground, position, mask=mask)
    else:
        result.paste(foreground, position)
    return result

def alpha_composite_images(bg_rgba, fg_rgba, position=(0, 0)):
    """RGBA 图像的 alpha 合成"""
    # 创建与背景同尺寸的画布
    canvas = bg_rgba.copy()
    
    # 创建临时画布用于合成
    tmp = Image.new('RGBA', bg_rgba.size, (0, 0, 0, 0))
    tmp.paste(fg_rgba, position)
    
    return Image.alpha_composite(canvas, tmp)

def create_collage(images, cols=2, padding=10, bg_color=(255, 255, 255)):
    """将多张图片拼接成网格"""
    n = len(images)
    rows = (n + cols - 1) // cols
    
    # 统一尺寸 (取最大宽高)
    max_w = max(img.width for img in images)
    max_h = max(img.height for img in images)
    
    cell_w = max_w + padding
    cell_h = max_h + padding
    
    collage_w = cols * cell_w + padding
    collage_h = rows * cell_h + padding
    
    collage = Image.new('RGB', (collage_w, collage_h), bg_color)
    
    for idx, img in enumerate(images):
        row = idx // cols
        col = idx % cols
        x = padding + col * cell_w
        y = padding + row * cell_h
        
        # 居中放置
        offset_x = x + (max_w - img.width) // 2
        offset_y = y + (max_h - img.height) // 2
        collage.paste(img, (offset_x, offset_y))
    
    return collage

# 测试
img1 = Image.fromarray(np.full((200, 200, 3), [255, 100, 100], dtype=np.uint8))
img2 = Image.fromarray(np.full((200, 200, 3), [100, 100, 255], dtype=np.uint8))

blended = blend_images(img1, img2, alpha=0.3)
print(f"混合: {np.array(blended).shape}")

collage = create_collage([img1, img2, blended, img1], cols=2)
print(f"拼接: 尺寸 {collage.size}")

# Alpha 合成测试
bg = Image.new('RGBA', (200, 200), (255, 255, 255, 255))
fg = Image.new('RGBA', (100, 100), (255, 0, 0, 128))  # 半透明红色
result = alpha_composite_images(bg, fg, position=(50, 50))
print(f"Alpha合成: 模式={result.mode}, 尺寸={result.size}")
```
