````python
# ====== 参考答案 ======

def histogram_equalization_solution(image):
    """
    手动实现直方图均衡化
    """
    # 1. 计算灰度直方图
    hist = np.zeros(256, dtype=np.int32)
    for pixel in image.ravel():
        hist[pixel] += 1
    
    # 也可以用 np.bincount
    # hist = np.bincount(image.ravel(), minlength=256)
    
    # 2. 计算累积分布函数 (CDF)
    cdf = hist.cumsum()
    
    # 3. 归一化CDF到 [0, 255]，构建查找表
    # 排除 cdf=0 的值
    cdf_min = cdf[cdf > 0].min()
    total_pixels = image.size
    
    # 映射公式: h(v) = round((cdf(v) - cdf_min) / (total_pixels - cdf_min) * 255)
    lut = np.zeros(256, dtype=np.uint8)
    for v in range(256):
        if cdf[v] > 0:
            lut[v] = np.round((cdf[v] - cdf_min) / (total_pixels - cdf_min) * 255)
    
    # 4. 应用查找表
    result = lut[image]
    
    return result, cdf, lut

# 测试
img = create_low_contrast_image()
result, cdf, lut = histogram_equalization_solution(img)

# 与 OpenCV 对比
result_cv = cv2.equalizeHist(img)

# 可视化
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# 原图和结果
axes[0][0].imshow(img, cmap='gray', vmin=0, vmax=255)
axes[0][0].set_title('原图（低对比度）')
axes[0][1].imshow(result, cmap='gray', vmin=0, vmax=255)
axes[0][1].set_title('手动均衡化结果')
axes[0][2].imshow(result_cv, cmap='gray', vmin=0, vmax=255)
axes[0][2].set_title('OpenCV 均衡化结果')

# 直方图对比
axes[1][0].hist(img.ravel(), 256, [0, 256], color='steelblue')
axes[1][0].set_title('原图直方图')
axes[1][0].set_xlim([0, 256])
axes[1][1].hist(result.ravel(), 256, [0, 256], color='coral')
axes[1][1].set_title('手动均衡化直方图')
axes[1][1].set_xlim([0, 256])
axes[1][2].hist(result_cv.ravel(), 256, [0, 256], color='green')
axes[1][2].set_title('OpenCV 均衡化直方图')
axes[1][2].set_xlim([0, 256])

for row in axes:
    for ax in row:
        ax.axis('off') if ax.images else None

plt.suptitle('直方图均衡化效果对比', fontsize=14)
plt.tight_layout()
plt.show()

print(f"原图像素值范围: [{img.min()}, {img.max()}]")
print(f"均衡化后范围: [{result.min()}, {result.max()}]")
print(f"与OpenCV结果差异: {np.abs(result.astype(int) - result_cv.astype(int)).mean():.2f} (平均像素差)")
````

````python
# ====== 参考答案 ======

def otsu_threshold_solution(image):
    """
    手动实现 Otsu 阈值算法
    """
    # 1. 计算灰度直方图
    hist = np.bincount(image.ravel(), minlength=256).astype(np.float64)
    total = image.size
    
    # 2. 遍历所有阈值，找最大类间方差
    best_thresh = 0
    max_var = 0
    between_class_vars = np.zeros(256)
    
    # 预计算
    sum_total = np.sum(np.arange(256) * hist)
    
    w0 = 0.0  # 前景比例
    sum0 = 0.0  # 前景加权和
    
    for t in range(256):
        w0 += hist[t]
        if w0 == 0:
            continue
        
        w1 = total - w0
        if w1 == 0:
            break
        
        sum0 += t * hist[t]
        
        mu0 = sum0 / w0
        mu1 = (sum_total - sum0) / w1
        
        var = (w0 / total) * (w1 / total) * (mu0 - mu1) ** 2
        between_class_vars[t] = var
        
        if var > max_var:
            max_var = var
            best_thresh = t
    
    # 3. 应用阈值
    _, binary_cv = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    binary = np.where(image > best_thresh, 255, 0).astype(np.uint8)
    
    return best_thresh, binary, between_class_vars

# 测试
img = create_bimodal_image()
thresh, binary, variances = otsu_threshold_solution(img)

# 与 OpenCV Otsu 对比
thresh_cv, binary_cv = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

print(f"手动 Otsu 阈值: {thresh}")
print(f"OpenCV Otsu 阈值: {thresh_cv}")
print(f"二值图一致性: {np.mean(binary == binary_cv)*100:.1f}%")

# 可视化
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0][0].imshow(img, cmap='gray')
axes[0][0].set_title('原图（双峰分布）')

axes[0][1].hist(img.ravel(), 256, [0, 256], color='steelblue')
axes[0][1].axvline(x=thresh, color='red', linestyle='--', label=f'Otsu阈值={thresh}')
axes[0][1].set_title(f'直方图 + Otsu阈值({thresh})')
axes[0][1].legend()

axes[1][0].imshow(binary, cmap='gray')
axes[1][0].set_title('手动 Otsu 分割结果')

axes[1][1].plot(variances)
axes[1][1].set_title('类间方差随阈值变化')
axes[1][1].set_xlabel('阈值')
axes[1][1].set_ylabel('类间方差')
axes[1][1].axvline(x=thresh, color='red', linestyle='--')

for row in axes:
    for ax in row:
        if ax.images:
            ax.axis('off')

plt.suptitle('Otsu 阈值分割', fontsize=14)
plt.tight_layout()
plt.show()
````

````python
# ====== 参考答案 ======

def template_matching_solution(scene, template, method=cv2.TM_CCOEFF_NORMED):
    """
    实现模板匹配
    """
    # 1. 模板匹配
    result = cv2.matchTemplate(scene, template, method)
    
    # 2. 找到最佳匹配位置
    min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)
    
    # 不同方法的最佳值位置不同
    if method in [cv2.TM_SQDIFF, cv2.TM_SQDIFF_NORMED]:
        top_left = min_loc
        val = min_val
    else:
        top_left = max_loc
        val = max_val
    
    # 3. 绘制匹配结果
    scene_color = cv2.cvtColor(scene, cv2.COLOR_GRAY2BGR)
    h, w = template.shape
    bottom_right = (top_left[0] + w, top_left[1] + h)
    cv2.rectangle(scene_color, top_left, bottom_right, (0, 0, 255), 2)
    
    return top_left, result, val, scene_color

# 测试
scene, template = create_scene_and_template()
top_left, result, val, scene_marked = template_matching_solution(scene, template)

print(f"模板大小: {template.shape}")
print(f"最佳匹配位置: {top_left}")
print(f"最佳匹配值: {val:.4f}")

# 可视化
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0][0].imshow(scene, cmap='gray')
axes[0][0].set_title('场景图像')
axes[0][1].imshow(template, cmap='gray')
axes[0][1].set_title('模板')
axes[1][0].imshow(result, cmap='hot')
axes[1][0].set_title(f'匹配响应图 (val={val:.4f})')
axes[1][0].plot(top_left[0], top_left[1], 'g+', markersize=15)
axes[1][1].imshow(cv2.cvtColor(scene_marked, cv2.COLOR_BGR2RGB))
axes[1][1].set_title('匹配结果（红框标记）')

for row in axes:
    for ax in row:
        ax.axis('off')
plt.tight_layout()
plt.show()

# 对比不同匹配方法
methods = [
    ('TM_SQDIFF', cv2.TM_SQDIFF),
    ('TM_SQDIFF_NORMED', cv2.TM_SQDIFF_NORMED),
    ('TM_CCORR', cv2.TM_CCORR),
    ('TM_CCORR_NORMED', cv2.TM_CCORR_NORMED),
    ('TM_CCOEFF', cv2.TM_CCOEFF),
    ('TM_CCOEFF_NORMED', cv2.TM_CCOEFF_NORMED),
]

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
for ax, (name, method) in zip(axes.ravel(), methods):
    top_left, result, val, marked = template_matching_solution(scene, template, method)
    ax.imshow(cv2.cvtColor(marked, cv2.COLOR_BGR2RGB))
    ax.set_title(f'{name}\nval={val:.4f}, pos={top_left}')
    ax.axis('off')
plt.suptitle('不同模板匹配方法对比', fontsize=14)
plt.tight_layout()
plt.show()
````

````python
# ====== 参考答案 ======

def connected_component_analysis_solution(image, min_area=50):
    """
    连通域分析
    """
    # 1. 连通域标记（8连通）
    num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(
        image, connectivity=8
    )
    
    print(f"检测到 {num_labels - 1} 个连通区域（不含背景）")
    print(f"{'ID':>4} {'面积':>6} {'位置(x,y)':>15} {'宽高':>12} {'质心':>15}")
    print("-" * 60)
    
    # 2. 生成彩色标记图，同时过滤小区域
    labeled = np.zeros((*image.shape, 3), dtype=np.uint8)
    colors = []
    
    np.random.seed(0)
    for i in range(num_labels):
        if i == 0:
            colors.append([0, 0, 0])  # 背景黑色
            continue
        
        area = stats[i, cv2.CC_STAT_AREA]
        x = stats[i, cv2.CC_STAT_LEFT]
        y = stats[i, cv2.CC_STAT_TOP]
        w = stats[i, cv2.CC_STAT_WIDTH]
        h = stats[i, cv2.CC_STAT_HEIGHT]
        cx, cy = centroids[i]
        
        if area >= min_area:
            color = np.random.randint(50, 255, 3).tolist()
            colors.append(color)
            print(f"{i:>4} {area:>6} ({x:>4},{y:>4})    {w:>3}x{h:<3}    ({cx:.1f}, {cy:.1f})")
            
            # 绘制边界框和质心
            labeled[labels == i] = color
        else:
            colors.append([0, 0, 0])  # 过滤掉的区域标为黑色
    
    # 3. 在标记图上绘制边界框和质心
    result = labeled.copy()
    for i in range(1, num_labels):
        area = stats[i, cv2.CC_STAT_AREA]
        if area < min_area:
            continue
        
        x = stats[i, cv2.CC_STAT_LEFT]
        y = stats[i, cv2.CC_STAT_TOP]
        w = stats[i, cv2.CC_STAT_WIDTH]
        h = stats[i, cv2.CC_STAT_HEIGHT]
        cx, cy = centroids[i]
        
        # 边界框
        cv2.rectangle(result, (x, y), (x + w, y + h), (255, 255, 255), 1)
        # 质心
        cv2.circle(result, (int(cx), int(cy)), 3, (255, 255, 255), -1)
        # 编号
        cv2.putText(result, str(i), (x, y - 5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255, 255, 255), 1)
    
    return result, stats, centroids, num_labels

# 测试
img = create_components_image()
result, stats, centroids, num = connected_component_analysis_solution(img, min_area=50)

# 可视化
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(img, cmap='gray')
axes[0].set_title('原图（二值图像）')

axes[1].imshow(result)
axes[1].set_title(f'连通域分析结果 ({num-1}个区域)')

# 面积柱状图
areas = []
labels_list = []
for i in range(1, num):
    area = stats[i, cv2.CC_STAT_AREA]
    if area >= 50:
        areas.append(area)
        labels_list.append(f'区域{i}')

if areas:
    axes[2].bar(labels_list, areas, color='steelblue')
    axes[2].set_title('各区域面积')
    axes[2].set_ylabel('面积（像素）')
    axes[2].tick_params(axis='x', rotation=45)
    for i, a in enumerate(areas):
        axes[2].text(i, a + 5, str(a), ha='center')

for ax in axes[:2]:
    ax.axis('off')
plt.tight_layout()
plt.show()
````
