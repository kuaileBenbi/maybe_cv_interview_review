```python
# ====== 参考答案 ======

def harris_corner_detection_solution(image, k=0.05, window_size=3, threshold=0.01):
    """
    手动实现 Harris 角点检测
    """
    # 1. 计算 x, y 方向梯度 (Sobel 算子)
    Ix = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
    Iy = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)
    
    # 2. 计算梯度乘积
    Ix2 = Ix ** 2
    Iy2 = Iy ** 2
    IxIy = Ix * Iy
    
    # 3. 高斯加权窗口求和
    Sx2 = cv2.GaussianBlur(Ix2, (window_size, window_size), sigmaX=1.5)
    Sy2 = cv2.GaussianBlur(Iy2, (window_size, window_size), sigmaX=1.5)
    Sxy = cv2.GaussianBlur(IxIy, (window_size, window_size), sigmaX=1.5)
    
    # 4. 计算 Harris 响应函数
    det_M = Sx2 * Sy2 - Sxy ** 2
    trace_M = Sx2 + Sy2
    R = det_M - k * (trace_M ** 2)
    
    # 5. 阈值筛选
    R_max = R.max()
    corner_mask = R > threshold * R_max
    
    # 6. 非极大值抑制 (膨胀后比较)
    dilated = cv2.dilate(R, np.ones((5, 5)))
    local_max = (R == dilated) & corner_mask
    
    corners = np.argwhere(local_max)  # (row, col) 格式
    
    return R, corners

# 测试
img = create_checkerboard()
R, corners = harris_corner_detection_solution(img)
print(f"检测到 {len(corners)} 个角点")

# 可视化
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 原图 + 角点
img_color = cv2.cvtColor(img.astype(np.uint8), cv2.COLOR_GRAY2BGR)
for pt in corners:
    cv2.circle(img_color, (pt[1], pt[0]), 3, (0, 0, 255), -1)
axes[0].imshow(cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB))
axes[0].set_title(f'Harris 角点检测结果 ({len(corners)}个)')

# 响应图
axes[1].imshow(R, cmap='hot')
axes[1].set_title('Harris 响应函数 R')
plt.colorbar(axes[1].images[0], ax=axes[1], fraction=0.046)
plt.tight_layout()
plt.show()
```

```python
# ====== 参考答案 ======

def sift_feature_matching_solution(img1, img2, ratio_thresh=0.75):
    """
    使用 SIFT + BFMatcher 进行特征匹配
    """
    # 1. 创建 SIFT 检测器
    sift = cv2.SIFT_create()
    
    # 2. 检测关键点并计算描述子
    kp1, des1 = sift.detectAndCompute(img1, None)
    kp2, des2 = sift.detectAndCompute(img2, None)
    
    print(f"图像1: {len(kp1)} 个关键点")
    print(f"图像2: {len(kp2)} 个关键点")
    
    # 3. 使用 BFMatcher 进行 KNN 匹配
    bf = cv2.BFMatcher(cv2.NORM_L2)
    raw_matches = bf.knnMatch(des1, des2, k=2)
    
    # 4. Lowe's ratio test 筛选
    good_matches = []
    for m_pair in raw_matches:
        if len(m_pair) == 2:
            m, n = m_pair
            if m.distance < ratio_thresh * n.distance:
                good_matches.append(m)
    
    return good_matches, kp1, kp2

# 测试
img1, img2 = create_test_images()
matches, kp1, kp2 = sift_feature_matching_solution(img1, img2)
print(f"\n筛选后好的匹配: {len(matches)} 对")

# 可视化匹配结果
match_img = cv2.drawMatches(
    img1, kp1, img2, kp2, matches[:50], None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)
plt.figure(figsize=(14, 5))
plt.imshow(match_img, cmap='gray')
plt.title(f'SIFT 特征匹配 (显示前50对, 共{len(matches)}对)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# ====== 参考答案 ======

def ransac_homography_solution(img1, img2, ratio_thresh=0.75, ransac_thresh=5.0):
    """
    使用 SIFT 匹配 + RANSAC 估计单应性矩阵
    """
    # 1. SIFT 特征提取与匹配
    sift = cv2.SIFT_create()
    kp1, des1 = sift.detectAndCompute(img1, None)
    kp2, des2 = sift.detectAndCompute(img2, None)
    
    bf = cv2.BFMatcher(cv2.NORM_L2)
    raw_matches = bf.knnMatch(des1, des2, k=2)
    
    # 2. Ratio test 筛选
    good_matches = []
    for m_pair in raw_matches:
        if len(m_pair) == 2:
            m, n = m_pair
            if m.distance < ratio_thresh * n.distance:
                good_matches.append(m)
    
    print(f"好的匹配: {len(good_matches)} 对")
    
    # 3. 提取匹配点坐标
    src_pts = np.float32([kp1[m.queryIdx].pt for m in good_matches]).reshape(-1, 1, 2)
    dst_pts = np.float32([kp2[m.trainIdx].pt for m in good_matches]).reshape(-1, 1, 2)
    
    # 4. RANSAC 估计单应性矩阵
    H, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, ransac_thresh)
    
    inliers = np.sum(mask)
    print(f"RANSAC 内点: {inliers} / {len(good_matches)}")
    
    return H, mask, good_matches, kp1, kp2

# 测试
img1, img2 = create_test_images()
H, mask, matches, kp1, kp2 = ransac_homography_solution(img1, img2)
print(f"\n单应性矩阵 H:\n{H}")

# 可视化内点匹配
inlier_matches = [m for m, mk in zip(matches, mask.ravel()) if mk]
match_img = cv2.drawMatches(
    img1, kp1, img2, kp2, inlier_matches[:30], None,
    matchColor=(0, 255, 0),
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)
plt.figure(figsize=(14, 5))
plt.imshow(match_img, cmap='gray')
plt.title(f'RANSAC 内点匹配 ({len(inlier_matches)}个内点)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# ====== 参考答案 ======

def compare_orb_sift_solution(img1, img2, ratio_thresh=0.75):
    """
    对比 ORB 和 SIFT 的速度和匹配效果
    """
    results = {}
    
    for name, detector in [('SIFT', cv2.SIFT_create()), ('ORB', cv2.ORB_create(nfeatures=2000))]:
        # 计时: 特征提取
        t0 = time.time()
        kp1, des1 = detector.detectAndCompute(img1, None)
        kp2, des2 = detector.detectAndCompute(img2, None)
        t_detect = time.time() - t0
        
        # 匹配
        norm_type = cv2.NORM_L2 if name == 'SIFT' else cv2.NORM_HAMMING
        bf = cv2.BFMatcher(norm_type)
        
        t0 = time.time()
        raw_matches = bf.knnMatch(des1, des2, k=2)
        
        # Ratio test
        good = []
        for m_pair in raw_matches:
            if len(m_pair) == 2:
                m, n = m_pair
                if m.distance < ratio_thresh * n.distance:
                    good.append(m)
        t_match = time.time() - t0
        
        results[name] = {
            'keypoints_img1': len(kp1),
            'keypoints_img2': len(kp2),
            'good_matches': len(good),
            'detection_time': t_detect,
            'match_time': t_match,
            'total_time': t_detect + t_match,
            'kp1': kp1, 'kp2': kp2, 'matches': good
        }
        
        print(f"--- {name} ---")
        print(f"  关键点: {len(kp1)} / {len(kp2)}")
        print(f"  好的匹配: {len(good)}")
        print(f"  特征提取耗时: {t_detect*1000:.1f} ms")
        print(f"  匹配耗时: {t_match*1000:.1f} ms")
        print(f"  总耗时: {(t_detect+t_match)*1000:.1f} ms")
        print()
    
    # 可视化对比
    fig, axes = plt.subplots(2, 1, figsize=(14, 10))
    
    for ax, name in zip(axes, ['SIFT', 'ORB']):
        r = results[name]
        match_img = cv2.drawMatches(
            img1, r['kp1'], img2, r['kp2'], r['matches'][:30], None,
            flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
        )
        ax.imshow(match_img, cmap='gray')
        ax.set_title(f"{name}: {r['good_matches']} matches, {r['total_time']*1000:.1f}ms")
        ax.axis('off')
    
    plt.tight_layout()
    plt.show()
    
    return results

# 测试
img1, img2 = create_test_images()
results = compare_orb_sift_solution(img1, img2)

# 速度对比柱状图
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
names = ['SIFT', 'ORB']

# 耗时对比
times = [results[n]['total_time']*1000 for n in names]
axes[0].bar(names, times, color=['#3498db', '#e74c3c'])
axes[0].set_ylabel('耗时 (ms)')
axes[0].set_title('总耗时对比')
for i, t in enumerate(times):
    axes[0].text(i, t+0.5, f'{t:.1f}ms', ha='center')

# 匹配数对比
matches_count = [results[n]['good_matches'] for n in names]
axes[1].bar(names, matches_count, color=['#3498db', '#e74c3c'])
axes[1].set_ylabel('好的匹配数')
axes[1].set_title('匹配质量对比')
for i, c in enumerate(matches_count):
    axes[1].text(i, c+0.5, str(c), ha='center')

plt.tight_layout()
plt.show()
```

## 三、面试要点总结

| 特征 | 尺度不变 | 旋转不变 | 描述子维度 | 速度 | 专利 |
|------|---------|---------|-----------|------|------|
| Harris | 否 | 是 | - | 快 | 否 |
| SIFT | 是 | 是 | 128 | 慢 | 已过期 |
| SURF | 是 | 是 | 64/128 | 中等 | 已过期 |
| ORB | 否(部分) | 是 | 256(二进制) | 很快 | 否 |

**关键面试题速记：**
- Harris：结构张量M，特征值判断，R = det(M) - k*trace(M)^2，旋转不变但尺度不变性不足
- SIFT：尺度空间 → DoG → 关键点 → 方向 → 128维描述子，全流程理解
- ORB：FAST + BRIEF，免费快速，适合实时应用
- RANSAC：随机采样一致，从噪声数据中鲁棒估计模型，剔除误匹配
