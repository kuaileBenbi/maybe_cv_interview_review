````python
# ====== 参考答案 ======

def manual_affine_transform_solution(image, angle=30, tx=50, ty=30, scale=0.8):
    """
    手动实现仿射变换：缩放 → 旋转 → 平移
    """
    h, w = image.shape[:2]
    cx, cy = w / 2, h / 2  # 旋转中心
    
    # 1. 构建变换矩阵（组合: 平移 * 旋转 * 缩放）
    theta = np.radians(angle)
    cos_t, sin_t = np.cos(theta), np.sin(theta)
    
    # 缩放矩阵
    S = np.array([[scale, 0, 0],
                  [0, scale, 0]], dtype=np.float64)
    
    # 旋转矩阵（绕中心）
    R = np.array([[cos_t, -sin_t, cx * (1 - cos_t) + cy * sin_t],
                  [sin_t,  cos_t, cy * (1 - cos_t) - cx * sin_t]], dtype=np.float64)
    
    # 组合: 先缩放，再旋转，再平移
    # M = Translation @ Rotation @ Scale (齐次坐标形式)
    # 简化为 2x3 矩阵
    T_mat = np.array([[1, 0, tx],
                      [0, 1, ty]], dtype=np.float64)
    
    # 组合矩阵
    RS = np.zeros((2, 3), dtype=np.float64)
    RS[:, :2] = R[:, :2] @ S[:, :2]  # 旋转 * 缩放
    RS[:, 2] = R[:, 2]  # 旋转的平移部分
    
    M = np.zeros((2, 3), dtype=np.float64)
    M[:, :2] = RS[:, :2]
    M[:, 2] = RS[:, 2] + T_mat[:, 2]  # 加上平移
    
    # 也可以直接用 cv2.getRotationMatrix2D 然后组合
    M_cv = cv2.getRotationMatrix2D((cx, cy), angle, scale)
    M_cv[:, 2] += [tx, ty]  # 加上额外平移
    
    # 2. 反向映射 + 双线性插值
    M_inv = np.vstack([M, [0, 0, 1]])  # 补成 3x3
    M_inv = np.linalg.inv(M_inv)[:2, :]  # 取前两行
    
    result = np.zeros_like(image)
    
    for y_out in range(h):
        for x_out in range(w):
            # 反向映射
            src = M_inv @ np.array([x_out, y_out, 1.0])
            x_src, y_src = src[0], src[1]
            
            # 边界检查
            if x_src < 0 or x_src >= w - 1 or y_src < 0 or y_src >= h - 1:
                continue
            
            # 双线性插值
            x0, y0 = int(x_src), int(y_src)
            dx, dy = x_src - x0, y_src - y0
            
            for c in range(image.shape[2]):
                val = (1 - dx) * (1 - dy) * image[y0, x0, c] + \
                      dx * (1 - dy) * image[y0, x0 + 1, c] + \
                      (1 - dx) * dy * image[y0 + 1, x0, c] + \
                      dx * dy * image[y0 + 1, x0 + 1, c]
                result[y_out, x_out, c] = int(val)
    
    return result, M

# 使用 OpenCV 的实现对比
img = create_test_image()
result, M = manual_affine_transform_solution(img, angle=30, tx=20, ty=15, scale=0.8)

# 用 cv2.warpAffine 做相同变换对比
cx, cy = img.shape[1] / 2, img.shape[0] / 2
M_cv = cv2.getRotationMatrix2D((cx, cy), 30, 0.8)
M_cv[:, 2] += [20, 15]
result_cv = cv2.warpAffine(img, M_cv, (img.shape[1], img.shape[0]))

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title('原图')
axes[1].imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
axes[1].set_title('手动实现（旋转30° + 平移 + 缩放0.8）')
axes[2].imshow(cv2.cvtColor(result_cv, cv2.COLOR_BGR2RGB))
axes[2].set_title('OpenCV 实现（对比）')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()
````

````python
# ====== 参考答案 ======

def perspective_transform_manual_solution(image, src_pts, dst_pts):
    """
    手动实现透视变换（DLT算法）
    """
    h, w = image.shape[:2]
    n = len(src_pts)
    
    # 1. 构建方程组 A*h = 0（DLT算法）
    A = np.zeros((2 * n, 9))
    for i in range(n):
        x, y = src_pts[i]
        xp, yp = dst_pts[i]
        A[2*i]     = [-x, -y, -1, 0, 0, 0, xp*x, xp*y, xp]
        A[2*i + 1] = [0, 0, 0, -x, -y, -1, yp*x, yp*y, yp]
    
    # 2. SVD 求解
    _, S, Vt = np.linalg.svd(A)
    h = Vt[-1, :]  # 最小奇异值对应的行
    
    # 3. reshape 为 3x3 矩阵并归一化
    H = h.reshape(3, 3)
    H = H / H[2, 2]  # 归一化使 H[2,2] = 1
    
    # 4. 应用变换（使用 cv2.warpPerspective）
    warped = cv2.warpPerspective(image, H, (w, h))
    
    return warped, H

# 测试
img = create_test_image()
h, w = img.shape[:2]

# 定义源点和目标点
src = np.float32([[50, 50], [250, 50], [250, 250], [50, 250]])
# 模拟一个倾斜视角
dst = np.float32([[100, 30], [270, 50], [240, 260], [60, 230]])

warped, H = perspective_transform_manual_solution(img, src, dst)

# 与 OpenCV 对比
H_cv = cv2.getPerspectiveTransform(src, dst)
warped_cv = cv2.warpPerspective(img, H_cv, (w, h))

print(f"手动计算的 H:\n{H}")
print(f"\nOpenCV 计算的 H:\n{H_cv}")
print(f"\n矩阵差异: {np.max(np.abs(H - H_cv)):.6f}")

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title('原图')
# 绘制源点
for pt in src:
    axes[0].plot(pt[0], pt[1], 'ro', markersize=8)

axes[1].imshow(cv2.cvtColor(warped, cv2.COLOR_BGR2RGB))
axes[1].set_title('手动透视变换')
axes[2].imshow(cv2.cvtColor(warped_cv, cv2.COLOR_BGR2RGB))
axes[2].set_title('OpenCV 透视变换')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()
````

````python
# ====== 参考答案 ======

def calibrate_camera_solution(images, board_size=(9, 6), square_size=25):
    """
    使用棋盘图像进行相机标定
    """
    cols, rows = board_size
    
    # 1. 准备物体点
    objp = np.zeros((cols * rows, 3), np.float32)
    objp[:, :2] = np.mgrid[0:cols, 0:rows].T.reshape(-1, 2)
    objp[:, :2] *= square_size
    
    obj_points = []  # 3D 点
    img_points = []  # 2D 点
    valid_images = []
    
    # 2. 在每张图像中查找棋盘角点
    for idx, img in enumerate(images):
        gray = img.copy()
        
        # 查找棋盘角点
        ret, corners = cv2.findChessboardCorners(gray, board_size, None)
        
        if ret:
            obj_points.append(objp)
            
            # 亚像素精化
            corners2 = cv2.cornerSubPix(
                gray, corners, (11, 11), (-1, -1),
                criteria=(cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)
            )
            img_points.append(corners2)
            valid_images.append(idx)
    
    print(f"成功检测角点的图像: {len(valid_images)} / {len(images)}")
    
    if len(valid_images) < 3:
        print("检测到的图像太少，无法标定")
        return None, None, None, None, None
    
    # 3. 相机标定
    img_size = (images[0].shape[1], images[0].shape[0])
    ret, cameraMatrix, distCoeffs, rvecs, tvecs = cv2.calibrateCamera(
        obj_points, img_points, img_size, None, None
    )
    
    # 4. 计算重投影误差
    total_error = 0
    total_points = 0
    for i in range(len(obj_points)):
        proj_points, _ = cv2.projectPoints(
            obj_points[i], rvecs[i], tvecs[i], cameraMatrix, distCoeffs
        )
        error = cv2.norm(img_points[i], proj_points, cv2.NORM_L2) / len(proj_points)
        total_error += error
        total_points += 1
    
    mean_error = total_error / total_points
    
    print(f"\n=== 标定结果 ===")
    print(f"内参矩阵 K:\n{cameraMatrix}")
    print(f"\n畸变系数: {distCoeffs.ravel()}")
    print(f"\n平均重投影误差: {mean_error:.4f} 像素")
    
    return cameraMatrix, distCoeffs, rvecs, tvecs, mean_error

# 生成标定图像并标定
images, _ = generate_chessboard_images(num_images=20)
cameraMatrix, distCoeffs, rvecs, tvecs, error = calibrate_camera_solution(images)

# 可视化部分检测到的棋盘角点
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
cols, rows = 9, 6
shown = 0
for idx, img in enumerate(images[:6]):
    ret, corners = cv2.findChessboardCorners(img, (cols, rows), None)
    ax = axes[shown // 3][shown % 3]
    if ret:
        vis = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
        cv2.drawChessboardCorners(vis, (cols, rows), corners, ret)
        ax.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
        ax.set_title(f'图像 {idx}: 检测成功')
    else:
        ax.imshow(img, cmap='gray')
        ax.set_title(f'图像 {idx}: 未检测到')
    ax.axis('off')
    shown += 1
plt.suptitle('棋盘角点检测结果', fontsize=14)
plt.tight_layout()
plt.show()
````

### 变换类型对比

| 变换类型 | 矩阵大小 | 自由度 | 保持性质 | 所需点对 |
|---------|---------|-------|---------|---------|
| 刚体变换 | 3x3(齐次) | 3 | 距离、角度 | 2 |
| 相似变换 | 3x3(齐次) | 4 | 角度、形状 | 2 |
| 仿射变换 | 2x3 | 6 | 平行性、比例 | 3 |
| 透视变换 | 3x3 | 8 | 直线性 | 4 |

### 相机模型核心公式

```
投影模型: s * [u,v,1]^T = K * [R|t] * [X,Y,Z,1]^T
内参矩阵: K = [[fx, 0, cx], [0, fy, cy], [0, 0, 1]]
畸变: 径向(k1,k2,k3) + 切向(p1,p2)
F = K^(-T) * E * K^(-1)
E = [t]_x * R
```

### 标定流程
1. 拍摄多张不同角度的棋盘图
2. 检测棋盘角点（findChessboardCorners + cornerSubPix）
3. 标定求解（calibrateCamera）
4. 评估重投影误差
5. 去畸变（undistort）
